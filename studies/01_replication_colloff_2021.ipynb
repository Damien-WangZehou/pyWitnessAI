## Similarity scores replication

A notebook to replicate the similarity scores generated by FRS (opencv+FaceNet) in 2022.

Please make sure you have the right version of pyWitnessAI installed

```
pip install -e ".[colloff_replication]"
```

The LineupLoader is a old class that used to load images from a directory and returns a list of images. It has been replaced by the Images class. Here we call it only for replication purposes.

````python
from pyWitnessAI import LineupLoader
````

````python
filler_pool = LineupLoader(directory_path="D:\PhD\Data\Colloff2021\Fillerpool") # Please use the absolute path
````

````python
filler_pool_face = filler_pool.load_images()
````

````python
target_faces = LineupLoader(directory_path="D:\PhD\Data\Colloff2021\TargetFace")
````

````python
target_faces_face = target_faces.load_image()
````

The default settings of `LineupLoader.compare_faces()`:
- model="Facenet"
- backend="opencv"
- distance_metric="euclidean"
- align=True
- enforce_detection=False
- normalization="base"

````python
similarity_scores = target_faces.compare_faces(target_faces_face, filler_pool_face)
````

````python
filler_pool.save(similarity_scores, label="replicate", column_name=target_faces.file_names)
````
