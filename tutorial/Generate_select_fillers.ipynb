```python
from pyWitnessAI import FaceGenerator
from IPython.display import Image, display
```

```python
generator = FaceGenerator(
    description=(
        "around 40 years old White male with short black hair, "
        "brown eyes, a beard, and an oval face"
    ),
    n=6,
    output_dir="./build/generated_faces",
)

results = generator.generate()
display(generator.to_manifest())
for face in results:
    display(Image(filename=face.path, width=240))
```
