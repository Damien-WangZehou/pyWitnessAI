# Installation Guide

## 1. Clone the Repository from GitHub

```bash
git clone https://github.com/Damien-WangZehou/pyWitnessAI.git
```

After downloading the repository, navigate to the project directory and follow these steps.


## 2. Create a Virtual Environment

Virtual environments help manage dependencies and avoid conflicts between different projects. Using them is highly recommended here as `pyWitnessAI` is using `mtcnn`, `retinaface`, and other deep learning models which have specific dependency requirements.

By creating the virtual environment, you can ensure the project runs smoothly with **reproducible results**.

You might need conda or venv to create a virtual environment. Here, we will use `conda`.

```bash
conda env create -f environment.yml
```

In environment.yml, we have specified all the required dependencies for `pyWitnessAI`. This includes libraries for deep learning, image processing, and other utilities.

Now you can activate the virtual environment:
```bash
conda activate pyWitnessAI
```

## * For developers


To update the dependencies, you can use:
```bash
conda env update -f environment.yml --prune
```
